```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('regularization_comparison_s26.csv')
print("imported")
df.head()
```

```python
# Your actual data
df = pd.DataFrame({
    'Model': [
        'Unregularized LR',
        'Ridge (L2)',
        'Lasso (L1)',
        'ElasticNet'
    ],
    'R2': [0.923820, 0.917973, 0.899018, 0.899018],
    'RMSE': [21.826064, 22.648182, 25.129109, 25.129109],
    'MAE': [15.009205, 15.157836, 17.207660, 17.207660]
})

# Create two plots
fig, ax = plt.subplots(1, 2, figsize=(14, 6))

# 1. Model Performance Plot
ax[0].scatter(
    df['R2'], df['RMSE'],
    alpha=0.8, edgecolors='k', color='dodgerblue', s=120
)

for i, model in enumerate(df['Model']):
    ax[0].annotate(
        model, (df['R2'][i], df['RMSE'][i]),
        xytext=(5, 5), textcoords='offset points', fontsize=8
    )

ax[0].set_xlabel('R² Score')
ax[0].set_ylabel('RMSE')
ax[0].set_title('Model Performance: R² vs RMSE')
ax[0].grid(True, linestyle=':', alpha=0.6)

# 2. RMSE vs MAE Comparison
x = np.arange(len(df))
width = 0.35

ax[1].bar(x - width/2, df['RMSE'], width, label='RMSE')
ax[1].bar(x + width/2, df['MAE'], width, label='MAE')

ax[1].set_xticks(x)
ax[1].set_xticklabels(df['Model'], rotation=25, ha='right')
ax[1].set_ylabel('Error Value')
ax[1].set_title('Error Comparison Across Models')
ax[1].legend()
ax[1].grid(axis='y', linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()

print("Model diagnostic comparison plots successfully rendered.")
```
