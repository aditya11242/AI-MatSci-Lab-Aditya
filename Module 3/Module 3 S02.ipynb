```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 1. Historical Lab Data
# (Carbon Percentage, Printing Temperature)
X_train = np.array([
    [0.1, 200],
    [0.2, 210],
    [0.4, 230],
    [0.5, 240],
    [0.6, 250],
    [0.8, 270]
])

# Measured Strength in MegaPascals (MPa)
# from past lab tests
y_train = np.array([
    300, 340, 410, 450, 490, 560
])

# 2. Training the Linear Regression Model
ai_model = LinearRegression()

ai_model.fit(X_train, y_train)

print("AI Training Complete! The model has learned your material data.")
```

```python
# 3. Interactive Parameters - Modify these to test new recipes
my_carbon_percentage = 0.25
my_printing_temp = 225

# 4. Running the Prediction Pipeline
new_material = np.array([[my_carbon_percentage, my_printing_temp]])

predicted_strength = ai_model.predict(new_material)

print(f"\nAI PREDICTION: {predicted_strength[0]:.2f} MPa")

# 5. Visualizing the Design Space
plt.figure(figsize=(6, 4))

plt.scatter(
    X_train[:, 0],
    y_train,
    color='blue',
    label='Past Lab Experiments',
    s=100
)

plt.scatter(
    my_carbon_percentage,
    predicted_strength,
    color='red',
    marker='*',
    label='Your Virtual AI Material',
    s=250
)

plt.xlabel('Carbon Percentage (%)')
plt.ylabel('Material Strength (MPa)')
plt.title('Materials Informatics: Lab Data vs AI Prediction')

plt.legend()
plt.grid(True)
plt.show()
```
