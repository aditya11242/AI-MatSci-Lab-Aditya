```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))
sns.histplot(df['Bandgap'], kde=True, bins=30)
plt.title('Distribution of Bandgap')
plt.xlabel('Bandgap')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['Formation_Energy'], kde=True, bins=30)
plt.title('Distribution of Formation Energy')
plt.xlabel('Formation Energy')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['Volume'], kde=True, bins=30)
plt.title('Distribution of Volume')
plt.xlabel('Volume')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(
    df[['Bandgap', 'Formation_Energy', 'Volume']].corr(),
    annot=True,
    cmap='coolwarm'
)
plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Formation_Energy', y='Bandgap')
plt.title('Bandgap vs Formation Energy')
plt.xlabel('Formation Energy')
plt.ylabel('Bandgap')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Volume', y='Bandgap')
plt.title('Bandgap vs Volume')
plt.xlabel('Volume')
plt.ylabel('Bandgap')
plt.show()
```

```python
sns.pairplot(df[['Bandgap', 'Formation_Energy', 'Volume']].dropna())
plt.show()
```
