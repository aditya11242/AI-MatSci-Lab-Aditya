## Quiz Question
**What is the difference between a Python List and a NumPy Array?**

A Python List can store elements of different data types and is flexible, but it is slower for mathematical operations. A NumPy Array requires all elements to be of the same data type, uses less memory, and supports fast, vectorized mathematical operations — making it much better for numerical and scientific computing.

```python
import matplotlib.pyplot as plt

# Data: Period 4 Transition Metals
elements = ['Titanium', 'Vanadium', 'Chromium', 'Manganese', 'Iron']
atomic_number = [22, 23, 24, 25, 26]
density = [4.506, 6.11, 7.15, 7.21, 7.874]  # g/cm^3

# Create the line plot
plt.figure(figsize=(8, 5))
plt.plot(atomic_number, density, marker='o', linestyle='-', color='blue')

# Label axes and title
plt.title('Atomic Number vs Density (Period 4 Transition Metals)')
plt.xlabel('Atomic Number')
plt.ylabel('Density (g/cm³)')
plt.xticks(atomic_number, elements, rotation=15)
plt.grid(True)
plt.tight_layout()

plt.show()
```
