```python
import pandas as pd
```

```python
# 1. Create a tiny dataset
data = {
    'Material': ['Iron', 'Alumina', 'Polyethylene'],
    'Class': ['Metal', 'Ceramic', 'Polymer'],
    'Melting_Temp_K': [1811, 2345, 400],
    'Density': [7.87, 3.95, 0.94]   # Density in g/cm³
}

df = pd.DataFrame(data)

# 2. Unit Conversion: Kelavin to Celsius
# Formula: C = K - 273.15
df['Melting_Temp_C'] = df['Melting_Temp_K'] - 273.15

print("--- After Unit Conversion ---")
print(df)

# 3. Categorical Encoding (One-Hot Encoding)
# Convert the 'Class' text column into binary columns
df_encoded = pd.get_dummies(df, columns=['Class'])

print("\n--- After Categorical Encoding ---")
print(df_encoded)

# 4. Function to filter materials by density
def filter_by_density(df, density):
    """
    Returns all materials having the specified density.
    """
    return df[df['Density'] == density]

# Example: Filter materials with density = 10 g/cm³
result = filter_by_density(df, 10)

print("\n--- Materials with Density = 10 g/cm³ ---")
if result.empty:
    print("No material found with density 10 g/cm³.")
else:
    print(result)
```
