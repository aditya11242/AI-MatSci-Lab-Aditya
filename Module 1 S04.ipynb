## Option 2: Thermal Conductivity Bar Chart

```python
import matplotlib.pyplot as plt

# Data: Thermal Conductivity (W/m·K)
metals = ['Aluminum', 'Copper', 'Silver', 'Gold']
conductivity = [237, 401, 429, 318]

# Create the bar chart
plt.figure(figsize=(8, 5))
plt.bar(metals, conductivity, color='green')

# Label axes and title
plt.title('Thermal Conductivity of Common Engineering Metals')
plt.xlabel('Metal')
plt.ylabel('Thermal Conductivity (W/m·K)')
plt.tight_layout()

plt.show()
```
